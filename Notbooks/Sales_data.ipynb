### Step 2 - Data Cleaning -

```python
# Remove dublicates
df = df.drop_duplicates()
df
```

```python
# find the missing values 
print(df.isnull().sum())  # This is not determine the empty colum of datetime - because NaT (Not a time) so ....
```

```python
df.loc[:, 'Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%Y', errors='coerce')

# This code converts your Order_Date column from text strings (like "2026-06-02" or "02/06/2026") 
# into official Python datetime objects so you can perform date calculations.
```

```python
df['Order_Date'].dtypes
```

```python
df = df.dropna(subset = ['Order_Date']) # delete row with NaT -Not a Time.
```

```python
df
```

```python
df['Profits'] = df['Sales'] - df['Cost']
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df
```

```python
df.to_csv('Sales_Data_Cleaned.csv',index=False) # remove index value in file
```

### # Python Data: - Excel -> SQL

```python
# import pandas and from sqlalchemy import create_engine
import pandas as pd
from sqlalchemy import create_engine
```

```python
# 1. Define your database connection engine
engine = create_engine('mysql+pymysql://root:root@localhost:3306/analytic_salesdb')
# make connection of mysql with python's mysql(pymysql)
#'mysql+pymysql://username:password@servername/databasename'
# Note : if use @ in your password then use, @ = %40, like admin@123 -> admin%40123
```

##### # Save your DataFrame back to a MySQL table

```python
df.to_sql(
    name='sales_data',  # table name as you wish
    con=engine,         # connection engine variable
    if_exists = 'replace',  # overight with current data
    index = False,       # avoid to create extra index column data
)
```
